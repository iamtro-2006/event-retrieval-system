```python
from pathlib import Path
import yaml

# ============================================================
# Config file trên Colab
# ============================================================

CONFIG_PATH = Path(
    "/content/event-retrieval-system/backend/configs/ocr_extraction.yaml"
)

# ============================================================
# Load config
# ============================================================

with open(CONFIG_PATH, "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

# ============================================================
# Sửa path cho cấu trúc Colab
#
# /content/
# ├── data/
# └── event-retrieval-system/
#     └── backend/
#         └── configs/
#             └── ocr_extraction.yaml
# ============================================================

config["dataset"]["root"] = "/content/data/ocr"

config["extraction"]["input_keyframes_root"] = \
    "/content/data/keyframes"

config["extraction"]["output_root"] = \
    "/content/data/ocr_test"

config["extraction"]["visualize_dir"] = \
    "/content/data/ocr_visuals"

# ============================================================
# Ghi ngược lại file YAML
# ============================================================

with open(CONFIG_PATH, "w", encoding="utf-8") as f:
    yaml.dump(
        config,
        f,
        allow_unicode=True,
        sort_keys=False,
        default_flow_style=False
    )

print(f"Updated: {CONFIG_PATH}")
```
